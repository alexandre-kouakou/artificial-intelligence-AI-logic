```python
class LinearRegression:
    def __init__(self):
        # placeholder for weights
        self.weights = None

    def fit(self, x_train, y_train):
        assert x_train.shape[0] == x_train.shape[0], "Number of samples in x and y should be equal"
        assert x_train.shape[0] > 0, "Training data cannot be empty"

        x_train = np.insert(x_train, 0, 1, axis = 1)
        self.weights = np.linalg.inv(x_train.T @ x_train) @ x_train.T @ y_train

    def predict(self, x_test):
        if self.weights is None:
            raise ValueError("Model has not been trained yet")

        x_test = np.insert(x_test, 0, 1, axis = 1)
        return x_test @ self.weights
```

```python
X, y = make_regression(n_samples = 100, n_features = 1, n_targets = 1, noise = 10)
```

```python
lr_model = LinearRegression()
lr_model.fit(X, y)
y_pred = lr_model.predict(X)
```

```python
plt.scatter(X, y, color = "blue", label = "Actual data")
plt.plot(X, y_pred, color = "red", label = "Predicted line")
plt.title("Linear Regression")
plt.show()
```
